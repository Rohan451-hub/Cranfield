```python
import math
```

```python
#Contantes
g = 9.81 #m/s2
R = 287 #J/kgK
L = 0.0065 #K/m
rho0 = 1.225 #kg/m3
T0 = 15 + 273.15 #K
P0 = rho0 * R * T0 #Pa
```

```python
H_Tropo = 11000
T_Tropo = T0 - L * H_Tropo
P_Tropo = P0 * (T_Tropo / T0) ** (g / (L * R))
```

```python
def compute_T(h):
    if h < H_Tropo:
        T = T0 - L * h
    else:
        T = T_Tropo
    return T
```

```python
def compute_P(h, T):
    if h < H_Tropo:
        P = P0 * (T / T0) ** (g / (L * R))
    else:
        P = P_Tropo * math.exp(-g * (h - H_Tropo) / (R * T))
    return P
```

```python
def compute_density(p, T):
    return p / (R * T)
```

```python
def ex1():
    headers = f"{'Altitude (m)':>12} | {'Temp (K)':>10} | {'Temp (°C)':>10} | {'Pression (Pa)':>14} | {'Masse Vol (kg/m³)':>18}"
    separator = "-" * len(headers)

    lines = [headers, separator]

    # Génération des données de 0 à 20 000 m par pas de 1 000 m
    for h in range(0, 20001, 1000):
        h_float = float(h)
        T = compute_T(h_float)
        p = compute_P(h_float, T)
        rho = compute_density(p, T)

        T_celsius = T - 273.15
        line = f"{h:>12} | {T:>10.2f} | {T_celsius:>10.2f} | {p:>14.2f} | {rho:>18.4f}"
        lines.append(line)

    output_text = "\n".join(lines)

    # Affichage dans la console
    print(output_text)

    # Sauvegarde dans un fichier texte
    file_name = "isa_atmosphere_table.txt"
    with open(file_name, "w", encoding="utf-8") as f:
        f.write(output_text)

    print(f"\nLe tableau a été sauvegardé avec succès dans le fichier '{file_name}'.")


if __name__ == "__main__":
    ex1()
```

```python
def compute_Thrust(thrust0, n, h):
    T_h = compute_T(h)
    P_h = compute_P(h, T_h)
    rho_h = compute_density(P_h, T_h)
    return thrust0 * (rho_h / rho0) ** n
```

```python
def ex2():
    thrust0 = 200000 #doesn't matter because canceled out during the 50% reduction
    n = 0.9
    target_thrust = 0.5 * thrust0
    altitude = 0
    step = 100
    i = 0
    thrust_computed = compute_Thrust(thrust0, n, altitude)
    while thrust_computed > target_thrust:
        altitude += step
        thrust_computed = compute_Thrust(thrust0, n, altitude)
        i += 1
    print(f"The altitude at which thrust is reduced by 50% : {altitude} m")
    print(f"The number of iterations at which thrust is reduced by 50% : {i}")
```

```python
if __name__ == "__main__":
    ex1()
    ex2()
```
